3 . Define Machine Learning. What are the main components in Machine Learning ?

Machine Learning (ML) is a branch of artificial intelligence (AI) that enables computers to learn from data and make decisions or predictions without being explicitly programmed for every task.


**Main Components of Machine Learning:**

1. Data

 - The most important ingredient.

 - Machines learn from examples (structured or unstructured data).

 - The better the data, the better the learning.

2. Model

 - A mathematical representation of a real-world process.

 - It tries to map inputs (features) to outputs (predictions or classifications).

3. Algorithm

 - The method used to train the model.

 - Examples: Linear Regression, Decision Trees, Neural Networks, etc.

 - It adjusts the model to fit the data.

4. Training

 - The process where the model learns patterns from data.

 - The algorithm optimizes the model by minimizing error during training.

5. Features

 - The input variables or characteristics used to make predictions.

 - Example: Age, Salary, and Experience could be features for predicting job performance.

6. Labels (or Targets)

 - The output the model tries to predict.

 - Example: In spam detection, the label is "spam" or "not spam".

7. Evaluation

 - After training, we test the model on new (test) data.

 - We use metrics like accuracy, precision, recall, or mean squared error to evaluate performance.

8. Prediction

 - Once trained and evaluated, the model is used to make predictions on new, unseen data.

5 . What are continuous and categorical variables ?

1. Continuous Variables
 - These are numerical values that can take any value within a range, often with decimals.

- Examples:
 - Height (e.g., 170.5 cm)

 - Weight (e.g., 65.3 kg)

 - Temperature (e.g., 36.6°C)

 - Salary (e.g., ₹50,000.75)

- Key Point:
 - You can measure them.

 - There are infinite possible values between two numbers.



2. Categorical Variables
 - These represent groups or categories, usually as names or labels. They tell you what type or which class something belongs to.

- Examples:
 - Gender (Male, Female, Other)

 - City (Delhi, Mumbai, Kolkata)

 - Color (Red, Blue, Green)

 - Yes/No answers

- Key Point:
 - You can count them, but not measure them.

 - They're often stored as text, but we can also encode them as numbers (like 0 for Male, 1 for Female).

6 . How do we handle categorical variables in Machine Learning? What are the common techniques ?

Categorical variables properly is super important for training a good machine learning model, because most ML algorithms only understand numbers, not text labels like "Male" or "Red".



1. Label Encoding
 - Assigns each category a unique number.

2. One-Hot Encoding
 - Creates a new binary column for each category.


3. Ordinal Encoding
 - Similar to label encoding, but you assign numbers with meaning/order.

4. Frequency Encoding
 - Replace categories with their frequency count or proportion in the dataset.

5. Target Encoding (Mean Encoding)
 - Replace each category with the average value of the target variable for that category.

16 . What is an Optimizer? What are different types of optimizers? Explain each with an example ?

- An optimizer is an algorithm used in machine learning to minimize or maximize an objective function, typically the loss function (which measures how well the model's predictions match the actual outcomes). Optimizers play a crucial role in training machine learning models by adjusting the model's parameters (e.g., weights in a neural network) during the training process, to reduce the error.

- The process of optimization involves updating the model's parameters based on the gradients of the loss function with respect to those parameters. This is often done iteratively using gradient descent or its variations.


image.png

1. Gradient Descent (GD)
 - Gradient Descent is a basic optimization algorithm. It involves updating the parameters (weights) of the model in the opposite direction of the gradient of the loss function with respect to the parameters.

 - Example: In linear regression, we minimize the mean squared error (MSE) between predicted and actual values by updating the model weights using gradient descent.

2. Stochastic Gradient Descent (SGD)
 - Stochastic Gradient Descent is a variation of gradient descent where instead of computing the gradient over the entire dataset, it computes the gradient using just one randomly selected data point (or example).

 - Example: For a classification problem like logistic regression, SGD can quickly adjust the weights based on random samples, helping the model converge faster.

3. Mini-batch Gradient Descent
 - Mini-batch Gradient Descent is a compromise between batch gradient descent (GD) and stochastic gradient descent (SGD). Instead of using all the data or just one data point, it uses a small subset (mini-batch) of the data for each update.

 - Example: In deep learning with large datasets, mini-batch gradient descent is typically used to update the parameters of neural networks.

4. Momentum Optimizer
 - Momentum is an enhancement to gradient descent that helps accelerate gradients in the correct direction, leading to faster convergence. It involves maintaining a running average of past gradients and updating the weights based on this moving average.

 - Example: In training a neural network, momentum helps speed up learning by adding velocity to the updates, especially when dealing with long narrow valleys (saddle points).

5. Nesterov Accelerated Gradient (NAG)
 - Nesterov Accelerated Gradient is a variant of momentum that anticipates the future gradient by looking ahead, rather than using only past gradients. This leads to faster convergence.

 - Example: NAG is often used in deep learning, where you need fast convergence and want to avoid getting stuck in local minima.

6. Adagrad (Adaptive Gradient Algorithm)
 - Adagrad adapts the learning rate based on the frequency of updates for each parameter. Rarely updated features get a larger learning rate, while frequently updated features get a smaller learning rate.

 - Example: Adagrad works well for sparse data (e.g., in NLP tasks), where some features are more important than others.


7. RMSprop (Root Mean Square Propagation)
 - RMSprop is an adaptive learning rate optimizer. It adjusts the learning rate for each parameter by using the moving average of squared gradients, which prevents the learning rate from decreasing too quickly.

 - Example: RMSprop is widely used in recurrent neural networks (RNNs) for tasks like language modeling.

8. Adam (Adaptive Moment Estimation)
 - Adam is one of the most popular optimizers, combining momentum and RMSprop. It computes adaptive learning rates for each parameter by using both the first moment (mean of gradients) and the second moment (variance of gradients).

 - Example: Adam is used in almost all types of neural network models, including image classification and language models.

21 . What is feature scaling? How does it help in Machine Learning ?

- Feature scaling refers to the process of standardizing or normalizing the range of independent variables (features) in a dataset. It ensures that each feature contributes equally to the machine learning model, especially in algorithms that rely on distance-based measurements, like K-Nearest Neighbors (KNN), Support Vector Machines (SVM), and Linear Regression.

- Feature scaling is important because many machine learning algorithms perform better or converge faster when the features are on a similar scale. Without scaling, features with larger numerical ranges can dominate the learning process, which can lead to suboptimal performance or incorrect results.
